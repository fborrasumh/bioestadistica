## 2. Conceptos clave (10 min)

### 2.1 Bioestadística

La **bioestadística** aplica los métodos estadísticos a las ciencias de la vida y de la salud. Sirve para describir lo que observamos, medir la incertidumbre y generalizar de una **muestra** de pacientes a la **población** de la que procede.

### 2.2 Variables clínicas

Cada característica que se registra de un paciente es una **variable**. Su tipo decide cómo se resume y qué prueba estadística se usa.

| Tipo | Subtipo | Ejemplos en el caso de Carmen |
| --- | --- | --- |
| Cuantitativa | Continua | Peso, talla, glucemia, presión arterial |
| Cuantitativa | Discreta | Número de ingresos en el último año |
| Cualitativa | Nominal | Fumadora (sí/no), sexo |
| Cualitativa | Ordinal | Grado de disnea mMRC (0 a 4) |

### 2.3 Por qué programar

Una hoja de cálculo sirve para 30 pacientes. Para 3.000, o para repetir el mismo análisis cada mes, el código es **reproducible** (cualquiera puede repetirlo), **escalable** (da igual el número de pacientes) y **auditable** (cada paso queda escrito).

### 3.2 Python como calculadora clínica

El **índice de masa corporal** (IMC) se calcula como el peso en kilogramos dividido entre el cuadrado de la talla en metros:

IMC = peso (kg) / talla (m)²

En Python, la potencia se escribe con `**`.

```python
68 / 1.63 ** 2
```

### 3.3 Variables con nombres clínicos

Guardar cada dato en una **variable** con un nombre claro, y con la unidad en el nombre, evita errores y hace el código legible. Lo que va detrás de `#` es un **comentario**: Python lo ignora, pero quien lee el código lo agradece.

```python
# Datos de la paciente Carmen (sintéticos)
peso_kg = 68
talla_m = 1.63     # en Python, el separador decimal es el punto

imc = peso_kg / talla_m ** 2

print(f"IMC de Carmen: {imc:.1f} kg/m²")
```

La expresión `f"...{imc:.1f}..."` es una **cadena formateada**: inserta el valor de `imc` redondeado a un decimal.

**Interpretación clínica:** según la clasificación de la OMS, un IMC entre 25,0 y 29,9 kg/m² corresponde a sobrepeso.

### 3.4 Comprobar un resultado

Antes de fiarnos de un cálculo, propio o del asistente, lo comprobamos con un valor conocido. La comparación `==` devuelve `True` (verdadero) o `False` (falso).

```python
round(imc, 1) == 25.6
```

### 3.5 Una mirada al destino del curso

La celda siguiente genera **200 pacientes sintéticos** y resume su presión arterial sistólica. **No hace falta entender este código todavía**: al final del Módulo 3 sabrás escribirlo tú.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(seed=1964)
pacientes = pd.DataFrame({
    "edad": rng.integers(40, 80, size=200),
    "fumador": rng.choice(["sí", "no"], size=200, p=[0.3, 0.7]),
    "pas_mmHg": rng.normal(135, 15, size=200).round(),
})

pacientes.groupby("fumador")["pas_mmHg"].describe().round(1)
```

## 4. Errores frecuentes (5 min)

**Usar la coma decimal.** En Python, `1,63` no es un número: es una pareja de valores (1 y 63). Ejecuta la celda y observa el resultado.

```python
talla_con_coma = 1,63
print(talla_con_coma, type(talla_con_coma))
```
